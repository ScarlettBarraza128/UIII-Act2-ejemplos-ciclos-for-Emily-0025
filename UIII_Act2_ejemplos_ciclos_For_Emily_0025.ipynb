```python
numero = int(input("Ingresa el número para el cual quieres la tabla de multiplicar: "))

# Recorre los números del 1 al 10
for i in range(1, 11):
    # Calcula el producto usando el operador aritmético *
    resultado = numero * i
    # Muestra el cálculo y el resultado en consola
    print(f"{numero} x {i} = {resultado}")
```

```python
start_num = int(input("Ingresa el número de inicio del rango: "))
end_num = int(input("Ingresa el número final del rango: "))

# Recorre el rango especificado por el usuario
for n in range(start_num, end_num + 1):
    # Verifica si el residuo es 0 (par) Y si es mayor a 10
    if (n % 2 == 0) and (n > 10):
        # Imprime solo los números que cumplen ambas condiciones
        print(f"El número {n} es par y mayor a 10")
```

```python
start_base = int(input("Ingresa el número de inicio para la base: "))
end_base = int(input("Ingresa el número final para la base: "))

start_exponente = int(input("Ingresa el número de inicio para el exponente: "))
end_exponente = int(input("Ingresa el número final para el exponente: "))

# Ciclo externo para la base
for base in range(start_base, end_base + 1):
    # Ciclo interno para el exponente
    for exponente in range(start_exponente, end_exponente + 1):
        # Calcula la potencia usando el operador aritmético **
        valor = base ** exponente
        # Imprime la relación entre ambos ciclos
        print(f"{base} elevado a {exponente} es {valor}")
```
